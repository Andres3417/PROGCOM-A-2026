```python

class Persona:

    def __init__(self, nombre, apellido, identificacion):
        self.nombre = nombre
        self.apellido = apellido
        self.identificacion = identificacion

    # Métodos para corregir datos
    def corregir_nombre(self, nuevo_nombre):
        self.nombre = nuevo_nombre

    def corregir_apellido(self, nuevo_apellido):
        self.apellido = nuevo_apellido

    # Mostrar información básica
    def mostrar_persona(self):
        print("Nombre:", self.nombre)
        print("Apellido:", self.apellido)
        print("ID:", self.identificacion)


# Clase hija
class Estudiante(Persona):

    def __init__(self, nombre, apellido, identificacion,
                 carrera, semestre):

        # Herencia
        super().__init__(nombre, apellido, identificacion)

        self.carrera = carrera
        self.semestre = semestre

    # Método corregir carrera
    def corregir_carrera(self, nueva_carrera):
        self.carrera = nueva_carrera

    # Imprimir información completa
    def imprimir_info(self):

        # Método heredado
        self.mostrar_persona()

        print("Carrera:", self.carrera)
        print("Semestre:", self.semestre)


# Clase Materia
class Materia:

    def __init__(self, nombre_materia, profesor, max_estudiantes):

        self.nombre_materia = nombre_materia
        self.profesor = profesor
        self.max_estudiantes = max_estudiantes

        # Lista de estudiantes
        self.estudiantes = []

    # Método matricular
    def matricular(self, estudiante):

        if len(self.estudiantes) < self.max_estudiantes:

            self.estudiantes.append(estudiante)

            print(estudiante.nombre,
                  "fue matriculado correctamente")

        else:
            print("No hay cupos disponibles")

    # Método nota
    def nota(self, estudiante, nota):

        print("La nota de",
              estudiante.nombre,
              "es:", nota)

    # Mostrar estudiantes
    def mostrar_estudiantes(self):

        print("\nEstudiantes matriculados:")

        for e in self.estudiantes:
            print(e.nombre, e.apellido)


# Programa principal

# Crear estudiantes
e1 = Estudiante(
    "Juan",
    "Pérez",
    101,
    "Ingeniería",
    3
)

e2 = Estudiante(
    "Ana",
    "Gómez",
    102,
    "Medicina",
    2
)

# Corregir datos
e1.corregir_nombre("Carlos")
e2.corregir_carrera("Derecho")

# Mostrar información
e1.imprimir_info()

print()

e2.imprimir_info()

# Crear materia
m1 = Materia(
    "Programación",
    "Profesor Luis",
    2
)

# Matricular estudiantes
m1.matricular(e1)
m1.matricular(e2)

# Mostrar estudiantes
m1.mostrar_estudiantes()

# Asignar notas
m1.nota(e1, 4.5)
m1.nota(e2, 3.8)
```
