```python
class Casa():
    # Constructor
    def __init__(self, paredes, color_paredes, techo, color_techo, plantas, ventanas):
        self.paredes = paredes
        self.color_paredes = color_paredes
        self.techo = techo
        self.color_techo = color_techo
        self.plantas = plantas
        self.ventanas = ventanas

    # Método para describir la casa
    def descripcion(self):
        print("Descripción de la casa:")
        print(f"La casa tiene {self.paredes} paredes de color {self.color_paredes}.")
        print(f"El techo es de {self.techo} color {self.color_techo}.")
        print(f"Hay {self.plantas} plantas en la puerta.")
        print(f"La casa tiene {self.ventanas} ventanas.")


# Crear objeto
mi_casa = Casa(4, "rosado", "ladrillo", "azul", "varias", 3)

# Mostrar descripción
mi_casa.descripcion()
```
