1. Es un paradigma de progamación, enfocado en construir objetos, (no orientado al procedimiento.)
2. tiene 4 pilares fundamentales:
herencia, polimorfismo, abstracción y encapsulamiento.

3. ventajas:
*Modularidad: Mejorar el código por trozos.
*Reusabilidad: Reusar clases y objetos.
*Mantenibilidad: Agregar fácilmente nuevos métodos.
*Escalabilidad: Escalar a nuevas aplicaciones desde un mismo código.
*Seguridad: No todos pueden alterar el código.

4. A los moldes los vamos a llamar clase.
5. lo que produce una case se llama instancia de clase o objeto.
6. Funciones dentro de clases se llaman métodos.
7. Las variables dentro de una clase se llaman atributos.

````python
class disfraz():
  #creando el consructor
  def __init__(self,color,talla,forma,material,personaje):
    #creando atributos de clase
    self.color = color
    self.talla = talla
    self.forma = forma
    self.material = material
    self.personaje = personaje
  def imprimir (self):
    return f"El disfraz es de {self.personaje}.\nEs de talla {self.talla} y de color {self.color} y de forma {self.forma}"
````

````python
perro = disfraz("morado","10","cuadrado","plastico","bob esponja")
print(perro.imprimir())
````
